```python
import matplotlib.pyplot as plt
import numpy as np

file_path = "/home/erdi/Uni/LCGAStage/Jets/LagrangianJet/Optimisaton/simu_opt_run_23_inj29.2500_dt0.0100_dist111_n10000/distrib_data.txt"

with open(file_path, "r") as f:
    line1 = f.readline().strip()
    line2 = f.readline().strip()

# Convert strings of numbers into numpy arrays (handles space, tab, or comma separation)
x = np.fromstring(line1.replace(",", " "), sep=" ")
y = np.fromstring(line2.replace(",", " "), sep=" ")

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, marker="o", markersize=3, linestyle="-", color="b", label="Data")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Distribution Data")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()

# Display or save
plt.show()
# plt.savefig("distribution_plot.png", dpi=300)
```
