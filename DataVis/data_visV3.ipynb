```python
# %matplotlib widget 
def plot_2d_hyperplanes(d, v):
    plt.figure(figsize=(8, 8))
    plt.scatter(d, v, s=1, alpha=0.6, color='blue')

    plt.title("2D Projection (d vs v)")
    plt.xlabel("List 'd'")
    plt.ylabel("List 'v'")
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()


def plot_3d_hyperplanes(d, v):
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    x = d[:-1]  
    y = v[:-1]  
    z = d[1:]   
    
    ax.scatter(x, y, z, s=1, alpha=0.6, color='red')
    
    ax.set_title("3D Projection (Marsaglia Effect)")
    ax.set_xlabel("d (i)")
    ax.set_ylabel("v (i)")
    ax.set_zlabel("d (i+1)")    
    ax.view_init(elev=15, azim=65)
    
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    file_d = "LCG_11mm.d"
    file_v = "LCG_11mm.v"
    
    d_data, v_data = LCG_11mm.d, LCG_11mm.v
    
    print("Opening 2D Plot. Close the window to view the 3D Plot...")
    plot_2d_hyperplanes(d_data, v_data)

    print("Opening 3D Plot. Click and drag to rotate the view and find the planes!")
    plot_3d_hyperplanes(d_data, v_data)
```
